Feature Engineering 

```python
Data['hour']=Data['transaction_time'].dt.hour
Data['day_of_week']=Data['transaction_time'].dt.day_name()
Data['month']=Data['transaction_time'].dt.month_name()
Data['is_weekend']=Data['transaction_time'].dt.dayofweek.isin([5,6]).astype(int)
```

```python
Data['Age']=((Data['transaction_time']-Data['dob']).dt.days//365).astype(int)
Data['Age_group']=pd.cut(Data['Age'],bins=[17,25,35,45,55,65,100],labels=['18-25', '26-35', '36-45', '46-55', '56-65', '65+'])
```

```python
Data['time_of_day']=pd.cut(Data['hour'],bins=[-1,5,7,11,17,23],labels=['Night','Morning','Afternoon','Evening','Late Night'])
```

```python
Data['amount_band']=pd.cut(Data['amount'],bins=[0,50,200,500,1000,Data['amount'].max()],labels=['<50', '50-200', '200-500', '500-1000', '1000+'])
```

```python
def haversine(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    a = np.sin((lat2 - lat1) / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2)**2
    return 6371 * 2 * np.arcsin(np.sqrt(a))

Data['distance_km'] = haversine(Data['customer_lat'], Data['customer_long'],
                              Data['merchant_lat'], Data['merchant_long'])
```

```python
import os
from sqlalchemy import create_engine


Data['merchant_id'] = Data['merchant'].factorize()[0] + 1

# Convert pd.cut columns from "categorical" to plain text so MySQL accepts them
for col in ['Age_group', 'time_of_day', 'amount_band']:
    Data[col] = Data[col].astype(str)

# Build the dimension tables
dim_customer = (Data[['card_number','gender','city','state','city_pop','customer_lat','customer_long']]
                  .drop_duplicates('card_number'))
dim_merchant = (Data[['merchant_id','merchant','category']]
                  .drop_duplicates('merchant_id')
                  .rename(columns={'merchant': 'merchant_name'}))
fact = Data[['transaction_id','card_number','merchant_id','transaction_time','amount','fraud_flag',
           'hour','day_of_week','month','is_weekend','Age','Age_group','time_of_day',
           'amount_band','distance_km','merchant_lat','merchant_long']]

print(len(dim_customer), len(dim_merchant), len(fact))   # expect 924, 693, 555719

# Connect - set MYSQL_USER and MYSQL_PASSWORD as environment variables before running
engine = create_engine(
    f"mysql+pymysql://{os.environ.get('MYSQL_USER', 'root')}:{os.environ['MYSQL_PASSWORD']}"
    "@localhost:3306/fraud_analysis")

# Load in this order: dimensions first, then the fact table
dim_customer.to_sql('dim_customer', engine, if_exists='append', index=False)
dim_merchant.to_sql('dim_merchant', engine, if_exists='append', index=False)
fact.to_sql('fact_transactions', engine, if_exists='append', index=False,
            chunksize=10000, method='multi')
```

```python
Data.to_parquet("../data/fraud_cleaned.parquet", index=False)
```

```python
from sqlalchemy import text

with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE fact_transactions"))

chunk = 10000
for i in range(0, len(fact), chunk):
    fact.iloc[i:i+chunk].to_sql('fact_transactions', engine, if_exists='append',
                                index=False, method='multi')
    print(f"{min(i + chunk, len(fact)):,} / {len(fact):,} rows loaded")
```
